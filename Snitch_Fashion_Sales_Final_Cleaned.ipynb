```python
import pandas as pd
import numpy as np

df = pd.read_csv('Snitch_Fashion_Sales_Uncleaned (1).csv')

print("Dataset Shape:", df.shape)
print("\nData Types:\n", df.dtypes)
print("\nMissing Values Before Cleaning:\n", df.isnull().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
display(df.head())
display(df.describe(include='all'))
```

```python
# Remove duplicate rows 
df.drop_duplicates(inplace=True)
display(df.head())
```

```python
#  Standardize City names 
city_mapping = {
    'bengaluru': 'Bangalore',
    'Hyd': 'Hyderabad',
    'hyderbad': 'Hyderabad'
}
df['City'] = df['City'].str.strip().replace(city_mapping)
display(df.head())
```

```python
#  Handle Units_Sold & flag Returned items (negative units)
df['Units_Sold'] = df['Units_Sold'].fillna(0).astype(int)
df['Units_Sold_Status'] = np.where(df['Units_Sold'] < 0, 'Returned', 'Delivered')
display(df.head())
```

```python
#  Impute missing Unit_Price using category-wise median
df['Unit_Price'] = df['Unit_Price'].fillna(
    df.groupby('Product_Category')['Unit_Price'].transform('median')
).round(2)
display(df.head())
```

```python
# Fix Discount_% anomalies (> 100%) and fill missing values with median (0.65)
discount_median = df['Discount_%'].median()  # 0.65
df.loc[df['Discount_%'] > 1.0, 'Discount_%'] = discount_median
df['Discount_%'] = df['Discount_%'].fillna(discount_median).round(2)
display(df.head())
```

```python
# Recalculate Sales_Amount accurately
df['Sales_Amount'] = ((df['Units_Sold'] * df['Unit_Price']) * (1 - df['Discount_%'])).round(2)
display(df.head())
```

```python
#  Standardize mixed Order_Date formats, extract Month/Year, and fill missing as 'Unknown'
parsed_dates = pd.to_datetime(df['Order_Date'], format='mixed', dayfirst=True, errors='coerce')
df['Order_Month'] = parsed_dates.dt.month_name().fillna('Unknown')
df['Order_Year'] = parsed_dates.dt.year.fillna(0).astype(int).astype(str).replace('0', 'Unknown')
df['Order_Date'] = parsed_dates.dt.strftime('%Y-%m-%d').fillna('Unknown')
display(df.head())
```

```python
# Fill missing Segment values with 'Unknown'
df['Segment'] = df['Segment'].fillna('Unknown')
display(df.head())
```

```python
# Categorical flags for EDA & Visualizations
df['Profit_Status'] = np.where(
    df['Profit'] > 0, 'Profitable',
    np.where(df['Profit'] < 0, 'Loss', 'Break-even')
)
df['Discount_Level'] = np.where(
    df['Discount_%'] > 0.50, 'High Discount (>50%)', 'Normal Discount (<=50%)'
)
display(df.head())
```

```python
# Define conditions and categories for Order Size
conditions = [
    (df['Units_Sold'] > 1),
    (df['Units_Sold'] == 1),
    (df['Units_Sold'] <= 0)
]

choices = ['Bulk Order', 'Single Order', 'Return / Zero']

# Create the new column using np.select
df['Order_Size'] = np.select(conditions, choices, default='Unknown')

# Verify the breakdown
print(df['Order_Size'].value_counts())
display(df.head())
```

```python
# Verify 0 missing values remain
print("Missing Values After Cleaning:\n", df.isnull().sum())
display(df.head())
```

```python
# Save the cleaned dataset
df.to_csv('Snitch_Fashion_Sales_Final_Cleaned.csv', index=False)
```
