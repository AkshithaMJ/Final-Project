## 🧵 Project Blueprint & Domain Context

### 🌟 Domain Landscape: Fashion Retail Analytics
In the fast-paced world of digital fashion retail (such as **Snitch Fashion**), business success relies heavily on balancing aggressive promotional discounts with healthy profit margins. Unlike standard retail, fashion sales are heavily driven by seasonal trends, product category popularity, and efficient inventory turnover, where managing returns and high discounts can directly dictate profitability.

### 📦 Dataset Anatomy
Our analytical canvas consists of a raw sales dataset featuring **2,500 transactions across 12 core dimensions**. Each row captures the pulse of a customer order—ranging from product categories and pricing structures to discount percentages, quantities sold, regional hubs, and net profitability.

### 🎯 Core Business Objectives
This exploratory data analysis is structured to uncover critical retail insights:
- **The Discount Paradox:** Evaluating whether massive promotional discounts (e.g., >50%) actually drive profitable volume or drain net revenue.
- **Fulfillment & Returns Dynamics:** Unpacking how order sizes and return statuses (`Units_Sold_Status`) impact overall business health.
- **Category Profitability:** Identifying which fashion lines act as revenue drivers versus margin detractors.

## 📦 1. Import Libraries & Load Dataset
Importing essential Python libraries (`pandas`, `numpy`) and loading the uncleaned dataset to inspect its initial state.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('Snitch_Fashion_Sales_Uncleaned (1).csv')

print("Dataset Shape:", df.shape)
print("\nData Types:\n", df.dtypes)
print("\nMissing Values Before Cleaning:\n", df.isnull().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
display(df.head())
display(df.describe())
```

## 🧹 2. Data Cleaning & Preprocessing

###  Remove Duplicate Rows
Checking and removing any duplicate records to ensure data integrity.

```python
df.drop_duplicates(inplace=True)
display(df.head())
```

###  Standardize City Names
Fixing spelling variations and inconsistencies in the `City` column (e.g., 'bengaluru' -> 'Bangalore', 'hyderbad' -> 'Hyderabad').

```python
city_mapping = {
    'bengaluru': 'Bangalore',
    'Hyd': 'Hyderabad',
    'hyderbad': 'Hyderabad'
}
df['City'] = df['City'].str.strip().replace(city_mapping)
display(df.head())
```

###  Handle Units Sold & Flag Returns
* Missing `Units_Sold` values are filled with `1`.
* Negative values in `Units_Sold` indicate returned items, so we create a new status column `Units_Sold_Status` ('Delivered' vs 'Returned').

```python
if 'Units_Sold' in df.columns:
    # Replace NaN and 0 values with 1
    df['Units_Sold'] = df['Units_Sold'].fillna(1)
    df['Units_Sold'] = df['Units_Sold'].replace(0, 1)
    
    # Convert to integer type
    df['Units_Sold'] = df['Units_Sold'].astype(int)
    
    print("Units_Sold value counts after cleaning:")
    print(df['Units_Sold'].value_counts())
    display(df.head())
```

###  Impute Missing Unit Prices
Filling missing `Unit_Price` values using the **median unit price** corresponding to each respective `Product_Category`.

```python
df['Unit_Price'] = df['Unit_Price'].fillna(
    df.groupby('Product_Category')['Unit_Price'].transform('median')
).round(2)
display(df.head())
```

### Clean and Impute Discounts
* Fixing discount anomalies where values exceeded 100% (`> 1.0`).
* Imputing missing discount percentages using the median value (`0.65`).

```python
discount_median = df['Discount_%'].median()  # 0.65
df.loc[df['Discount_%'] > 1.0, 'Discount_%'] = discount_median
df['Discount_%'] = df['Discount_%'].fillna(discount_median).round(2)
display(df.head())
```

### Recalculate Sales Amount
Accurately recalculating `Sales_Amount` using the formula:  
$$\text{Sales Amount} = (\text{Units Sold} \times \text{Unit Price}) \times (1 - \text{Discount }\%)$$

```python
df['Sales_Amount'] = ((df['Units_Sold'] * df['Unit_Price']) * (1 - df['Discount_%'])).round(2)
display(df.head())
```

###  Parse Dates and Fill Missing Values
Standardizing mixed date formats, using Forward Fill (`ffill`) for missing dates, and extracting Month/Year features.

```python
if 'Order_Date' in df.columns:
    parsed_dates = pd.to_datetime(df['Order_Date'], format='mixed', dayfirst=True, errors='coerce')
    df['Order_Date'] = parsed_dates
    df['Order_Date'] = df['Order_Date'].ffill() 
    
    # Forward fill missing dates using previous valid date
    
    df['Order_Month'] = df['Order_Date'].dt.month_name()
    df['Order_Year'] = df['Order_Date'].dt.year.astype(int).astype(str)
    df['Order_Date'] = df['Order_Date'].dt.strftime('%Y-%m-%d')
    display(df.head())
```

###  Handle Missing Segment Values
Filling missing values in the `Segment` column with 'Unknown' to ensure categorical consistency across the dataset.

```python
df['Segment'] = df['Segment'].fillna('Unknown')
display(df.head())
```

## ⚙️ 3. Feature Engineering & Categorical Flagging
Creating categorical flags and derived columns (`Profit_Status` and `Discount_Level`) to help with deeper business insights and visualizations.

```python
df['Profit_Status'] = np.where(
    df['Profit'] > 0, 'Profitable',
    np.where(df['Profit'] < 0, 'Loss', 'Break-even')
)
df['Discount_Level'] = np.where(
    df['Discount_%'] > 0.50, 'High Discount (>50%)', 'Normal Discount (<=50%)'
)
display(df.head())
```

###  Define Order Size Categories
Classifying orders into `Bulk Order`, `Single Order`, or `Return / Zero` based on the number of units sold using `np.select`.

```python
conditions = [
    (df['Units_Sold'] > 1),
    (df['Units_Sold'] == 1),
    (df['Units_Sold'] <= 0)
]

choices = ['Bulk Order', 'Single Order', 'Return / Zero']

# Create the new column using np.select
df['Order_Size'] = np.select(conditions, choices, default='Unknown')

# Verify the breakdown
print(df['Order_Size'].value_counts())
display(df.head())
```

## 📉 4. Outlier Detection & Handling (All Numerical Columns)
Applying the Interquartile Range (IQR) method across all numerical columns (`Sales_Amount`, `Units_Sold`, `Unit_Price`, `Discount_%`) to handle extreme values.

```python
# List of numerical columns to check for outliers
numerical_cols = ['Sales_Amount', 'Units_Sold', 'Unit_Price', 'Discount_%']

for col in numerical_cols:
    if col in df.columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        # Count outliers for each column
        outliers_count = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]
        print(f"Outliers detected in {col}: {outliers_count}")

        # Cap the outliers to retain data distribution
        df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)

print("\nOutlier handling completed for all numerical columns!")
display(df.head())
```

## Step 5: Final Verification & Preview
Checking for any remaining missing values across all columns to ensure data cleanliness and previewing the final dataset.

```python
# Verify 0 missing values remain
print("Missing Values After Cleaning:\n", df.isnull().sum())
display(df.head())
```
